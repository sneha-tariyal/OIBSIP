## Handling Missing Values

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage
```

```python
# Missing Description values were replaced with Unknown, while rows with missing CustomerID were removed because the customer information was unavailable.

df['Description'] = df['Description'].fillna('Unknown')
```

```python
df= df.dropna(subset=['CustomerID'])
```

## Duplicate remove

```python
# Duplicate rows were identified and removed to prevent from affecting the analysis

df = df.drop_duplicates()
```

## Outlier Detection Using IQR Method

```python
for col in ['Quantity','UnitPrice']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3-Q1
    lower = Q1-1.5*IQR
    upper = Q3+1.5*IQR

    outliers= df[(df[col]<lower)| (df[col]>upper)]

    print(f'\n{col}')
    print('Lower limit:',lower)
    print('Upper limit:',upper)
    print('Number of outliers:',len(outliers))
```

### Outlier Detection and Handling

The IQR method identified 25,616 potential outliers in Quantity and 34,112 in UnitPrice. These values were identified as potential outliers and were not directly removed using the IQR method. Separately, rows with non-positive UnitPrice and non-positive Quantity were removed during data cleaning to exclude these records from the final dataset. This may exclude some genuine product returns or cancellations represented by negative quantities.
